# Introduction
In this lesson, you will learn how to use different types of operators in Java to write expressions and compare values. Operators are symbols that tell the computer what kind of operation to do with the values in your code.

## Key Terms
- **Binary operator** – Uses two values (like `3 + 2`). Common for math operations such as `+`, `-`, `*`, `/`, and `%`.
- **Unary operator** – Uses one value (like `-4` or `++x`). These operators can change or check a single value.
- **Assignment operator** – Assigns a value to a variable (like `x = 5`). You’ll also use special versions like `+=` or `-=`.
- **Relational operator** – Compares two values (like `x > 3`) and returns `true` or `false`.
- **Order of precedence** – The rules Java follows to decide which operation to do first when there are multiple operators in an expression.


# Binary Operators

Binary operators (like `+`, `-`, `*`, `/`, and `%`) are called “binary” because they work with two values. You place one value on each side of the operator.  

For example:
4 + 2 → The operator + is between two values: 4 and 2.


---

## In Java, binary operators include:

- **Addition (`+`)** – Adds two values  
  → `2 + 3` results in `5`

- **Subtraction (`-`)** – Subtracts the second value from the first  
  → `7 - 2` results in `5`

- **Multiplication (`*`)** – Multiplies two values  
  → `4 * 3` results in `12`

- **Division (`/`)** – Divides the first value by the second  
  → `10 / 2` results in `5`

- **Modulus (`%`)** – Finds the remainder when one number is divided by another  
  → `5 % 2` results in `1` (because 5 ÷ 2 is 2 with remainder 1)

In Java, all of these are used between two numbers (operands).
