# Introduction
In this lesson, you will learn how to use different types of operators in Java to write expressions and compare values. Operators are symbols that tell the computer what kind of operation to do with the values in your code.

## Key Terms
- **Binary operator** – Uses two values (like `3 + 2`). Common for math operations such as `+`, `-`, `*`, `/`, and `%`.
- **Unary operator** – Uses one value (like `-4` or `++x`). These operators can change or check a single value.
- **Assignment operator** – Assigns a value to a variable (like `x = 5`). You’ll also use special versions like `+=` or `-=`.
- **Relational operator** – Compares two values (like `x > 3`) and returns `true` or `false`.
- **Order of precedence** – The rules Java follows to decide which operation to do first when there are multiple operators in an expression.


# Binary Operators

Binary operators (like `+`, `-`, `*`, `/`, and `%`) are called “binary” because they work with two values. You place one value on each side of the operator.  

For example:
4 + 2 → The operator + is between two values: 4 and 2.


---

## In Java, binary operators include:

- **Addition (`+`)** – Adds two values  
  → `2 + 3` results in `5`

- **Subtraction (`-`)** – Subtracts the second value from the first  
  → `7 - 2` results in `5`

- **Multiplication (`*`)** – Multiplies two values  
  → `4 * 3` results in `12`

- **Division (`/`)** – Divides the first value by the second  
  → `10 / 2` results in `5`

- **Modulus (`%`)** – Finds the remainder when one number is divided by another  
  → `5 % 2` results in `1` (because 5 ÷ 2 is 2 with remainder 1)

In Java, all of these are used between two numbers (operands).

---

```java
int sum = 4 + 3;       // sum is 7
int product = 2 * 5;   // product is 10
int mod = 9 % 4;       // mod is 1



## Popcorn Hack 1

Why are binary operators 'binary'? Provide 3 examples of a binary operator.

# Practice

Read each question and select the correct answer choice for each.  

---

### Question 1
What is the result of `7 % 3`?

- [ ] 3  
- [ ] 0  
- [ ] 1  
- [ ] 2  

**Check**

---

### Question 2
Fill in the missing number.  

What is `12 / 4`?  

**Answer:** _______  

**Check**

---

### Question 3
True or False: `6 * 2 = 12`

- [ ] False  
- [ ] True  

**Check**

---

## Solutions

<details>
<summary>Click to reveal answers</summary>

- **Question 1:** `7 % 3 = 1`  
- **Question 2:** `12 / 4 = 3`  
- **Question 3:** `True` (because `6 * 2` equals `12`)  

</details>



# Output

In the previous section, we explored **expressions** in Java – combinations of variables, values, and operators that evaluate to a single result. However, expressions alone do not help us **see** the results of our computations. That’s where **output** comes in.

---

## What is Output?
In programming, **output** is how a program communicates information back to the user.  
In Java, the most common way to produce text output is through the `System.out` object, which represents the **standard output stream** (usually the terminal/console).

---

### The `System.out.print` and `System.out.println` Methods

- **`System.out.print()`**  
  Prints text or values **without** moving to a new line afterwards.  
  The cursor stays on the same line.

- **`System.out.println()`**  
  Prints text or values **and** then moves the cursor to a **new line**.  
  (The `ln` at the end stands for **line**.)

### Example
System.out.print("Hello");
System.out.print("World");

**Output:**
HelloWorld





System.out.println("Hello");
System.out.println("World");




## 1. Menu Integration

Let’s look at how **`Menu.java` itself uses output**. 

```java
public class Menu {
    public static void main(String[] args) {
        System.out.println("Welcome to the Menu!");
        System.out.println("Choose an option:");
        
        MenuRow[] rows = new MenuRow[]{
            new MenuRow("Exit", () -> main(null)),
            new MenuRow("Output Lesson", () -> OutputLesson.main(null)),
            new MenuRow("Do Nothing", () -> DoNothingByValue.main(null)),
            new MenuRow("Swap if Hi-Low", () -> IntByReference.main(null)),
            new MenuRow("Matrix Reverse", () -> Matrix.main(null)),
            new MenuRow("Diverse Array", () -> DiverseArray.main(null)),
            new MenuRow("Random Squirrels", () -> Number.main(null))
        };

        for (int i = 0; i < rows.length; i++) {
            System.out.println((i + 1) + ": " + rows[i].getTitle());
        }
    }
}





Notice how System.out.println is used to:

Greet the user ("Welcome to the Menu!")

Print instructions ("Choose an option:")

Print each menu row as output (loop with numbers and labels).

The menu itself is output — it prints text and options for the user to see.

## OutputLesson.java (Focused Examples)

Now we’ll create a separate file for experimenting with output.

```java
package com.open.spring.hacks.methodsDataTypes;

public class OutputLesson {
    public static void main(String[] args) {
        System.out.println("=== Output Lesson ===");

        // 1. println vs print
        System.out.println("\n--- println vs print ---");
        System.out.println("Hello, world!"); // prints on a line by itself
        System.out.print("Hello, ");        // stays on the same line
        System.out.print("world!");         // continues same line
        System.out.println("  <-- same line above");

        // 2. Printing variables
        System.out.println("\n--- Printing variables ---");
        int x = 5;
        int y = 3;
        System.out.println("The sum is " + (x + y)); // Correct: 8
        System.out.println("Concatenation (no parentheses): " + x + y); // 53

        // 3. Escape sequences
        System.out.println("\n--- Escape sequences ---");
        System.out.println("Hello\nWorld");       // newline
        System.out.println("Name:\tJohn");        // tab
        System.out.println("She said, \"Hi!\""); // quotes
        System.out.println("C:\\Users\\Admin");   // backslash

        // 4. Expressions directly inside output
        System.out.println("\n--- Expressions in output ---");
        System.out.println("2 + 3 = " + (2 + 3));
        System.out.println("10 / 4 = " + (10 / 4));     // integer division
        System.out.println("10 / 4.0 = " + (10 / 4.0)); // double division
        System.out.println("5 % 2 = " + (5 % 2));
    }
}


## Types of Output (Deep Dive)  

### `System.out.println()`  
- Prints whatever you give it.  
- Moves to the next line **after** printing.  
- Good for separating outputs.  

### `System.out.print()`  
- Prints text **without** moving to the next line.  
- The next print will continue on the **same line**.  

### `System.out.printf()` (Advanced — Optional)  
- Allows **formatted output** with placeholders.  
- Example:  
  ```java
  System.out.printf("Pi approx = %.2f\n", Math.PI);


## Key Takeaways  

- **Output = communication from program to user.**  
- `System.out.println` → prints with a newline.  
- `System.out.print` → prints without newline.  
- **Escape sequences** let us control how text appears (`\n`, `\t`, `\"`, `\\`).  
- You can output **variables, calculations, and strings together** using `+`.  
- **Menu.java** itself is a real-world example of output in action.
