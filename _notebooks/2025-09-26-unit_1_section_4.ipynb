The goal of this lesson is to teach you two key concepts:

1. How to assign values to variables

2. How to take in user input and use them in your program

### Assignment Statements

Formatting for assigning a variable: `[data-type] [variable-name] = [variable-value];`

Below is an example of the assignment of different data types:

```
public class Main {
    public static void main(String[] args) {
         // Assign a String
        String name = "Billy";

        // Assign an Integer
        int age = 10;

        // Assign a Double
        double height = 165.34;

        // Assign a Char
        char firstLetter = 'B';

        // Assign a Boolean
        boolean isStudent = true;
    }
}
```
