## Assignment Statements and Input

### Assignment Statements

```
public class Main {
    public static void main(String[] args) {
         // Assign a String
        String name = "Billy";

        // Assign an Integer
        int age = 10;

        // Assign a Double
        double height = 165.34;

        // Assign a Char
        char firstLetter = 'B';

        // Assign a Boolean
        boolean isStudent = true;
    }
}
```
