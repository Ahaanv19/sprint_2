# CSA Unit 1.4 Lesson: Assignment Statements and Input

## Learning Objectives
- **LO 1.4.A**: Develop code for assignment statements with expressions and determine the value stored in variables.  
- **LO 1.4.B**: Develop code to read input.  

## Essential Knowledge
- **1.4.A.1**: Every variable must be assigned a value before use. Must come from a compatible data type. Variables are initialized the first time they are assigned. Reference types can be assigned an object or `null`.  
- **1.4.A.2**: The assignment operator `=` stores the value of an expression on the right into the variable on the left.  
- **1.4.A.3**: During execution, an expression is evaluated into a single value.  
- **1.4.B.1**: Input can be obtained in Java using the `Scanner` class (keyboard input).  

## 1. Assignment Statements

Variables must be **declared** with a type and then **assigned** a value.

````java
// Example 1: Assigning Integers
int a = 5;
int b = 3;
int c = a + b;  // c = 8
````

````java
// Example 2: Reassignment
int x = 10;
x = 20;   // previous value replaced
````

````java
// Example 3: Reference and null
String name = null;        // no object assigned
name = "Ahaan";            // now it refers to "Ahaan"
````

## 2. Input with Scanner

The `Scanner` class is used to read user input from the keyboard.

````java
import java.util.Scanner;

Scanner sc = new Scanner(System.in);
System.out.print("Enter your name: ");
String name = sc.nextLine();
System.out.println("Hello, " + name + "!");
````

````java
System.out.print("Enter a number: ");
int num = sc.nextInt();
System.out.println("You entered " + num);
````

## 4. InputLesson.java

Here is the full implementation of the Input lesson:

````java
package com.open.spring.hacks.methodsDataTypes;
import java.util.Scanner;

public class InputLesson {
    public static void main(String[] args) {
        System.out.println("=== Input Lesson ===");

        Scanner sc = new Scanner(System.in);

        // Example 1: String input
        System.out.print("Enter your name: ");
        String name = sc.nextLine();
        System.out.println("Hello, " + name + "!");

        // Example 2: Integer input
        System.out.print("Enter your age: ");
        int age = sc.nextInt();
        System.out.println("Next year you will be " + (age + 1));

        // Example 3: Double input
        System.out.print("Enter a decimal number: ");
        double d = sc.nextDouble();
        System.out.println("Half of that is " + (d / 2));
    }
}
````

    function updateUI() {
      document.getElementById("lives").innerHTML = "❤️".repeat(lives);
      document.getElementById("score").innerText = "Score: " + score;
      let progress = ((current) / quiz.length) * 100;

    }

    function loadQuestion() {
      if (current >= quiz.length || lives <= 0) {
        endGame();
        return;
      }
      document.getElementById("question").innerText = quiz[current].q;
      let choicesDiv = document.getElementById("choices");
      choicesDiv.innerHTML = "";
      quiz[current].choices.forEach((choice, i) => {
        let btn = document.createElement("button");
        btn.innerText = choice;
        btn.onclick = () => checkAnswer(i);
        choicesDiv.appendChild(btn);
      });
      updateUI();
    }

    function checkAnswer(selected) {
      if (selected === quiz[current].answer) {
        score++;
        alert("✅ Correct!");
      } else {
        lives--;
        alert("❌ Wrong! Lost a life.");
      }
      current++;
      loadQuestion();
    }

    function endGame() {
      document.getElementById("question").innerText = "";
      document.getElementById("choices").innerHTML = "";
      document.getElementById("end").innerText =
        lives > 0 ? "🎉 You Win! Final Score: " + score : "💀 Game Over! Final Score: " + score;
      document.getElementById("bar").style.width = "100%";
    }

    // start game
    loadQuestion();
